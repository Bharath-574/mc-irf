```python
# Cell 2: Download GGUF Models
import subprocess
from pathlib import Path

models_dir = Path("/home/bharath/Research/models")
models_dir.mkdir(parents=True, exist_ok=True)

gguf_models = [
    ("microsoft/Phi-3.5-mini-instruct-gguf", "Phi-3.5-mini-instruct-Q4_K_M.gguf", "phi-3.5-mini-gguf"),
    ("Qwen/Qwen2.5-7B-Instruct-GGUF", "Qwen2.5-7B-Instruct-Q4_K_M.gguf", "qwen2.5-7b-gguf"),
]

for repo, file, local_name in gguf_models:
    local_dir = models_dir / local_name
    local_dir.mkdir(exist_ok=True)
    gguf_path = local_dir / file
    
    if gguf_path.exists():
        print(f"✅ {file} already exists at {gguf_path}")
    else:
        print(f"⬇️ Downloading {file}...")
        result = subprocess.run([
            "huggingface-cli", "download", repo, file,
            "--local-dir", str(local_dir)
        ], capture_output=True, text=True)
        if result.returncode == 0:
            print(f"✅ Downloaded {file}")
        else:
            print(f"❌ Failed: {result.stderr}")
```

```python
# Cell 3: Generate Test Problems
from mc_irf.data.gsm_symbolic import generate_problem_set, CORE_TEMPLATES, save_problems

print("Generating GSM-Symbolic + NoOp problems...")

problems = generate_problem_set(
    templates=CORE_TEMPLATES,
    num_problems=20,
    seeds=[42, 123, 456],
    nop_types=["quantity", "comparison", "filter", "discount"],
)

print(f"Generated {len(problems)} problems")
clean = [p for p in problems if p.nop_type is None]
print(f"  Clean: {len(clean)}")
for nt in ["quantity", "comparison", "filter", "discount"]:
    print(f"  {nt}: {len([p for p in problems if p.nop_type == nt])}")

# Save for experiments
save_problems(problems, "data/processed/local_test_problems.jsonl")
print("Saved to data/processed/local_test_problems.jsonl")
```

```python
# Cell 4: Run E1 - NoOp Persistence Baseline
from mc_irf.experiments.runner import run_experiment

print("Running E1: NoOp Persistence Baseline (local)")
print("Generator: phi-3.5-mini-gguf, No discriminator")

result = run_experiment("E1", environment="local")

print(f"\n=== E1 Complete ===")
print(f"Generator Accuracy: {result.gen_acc:.3f}")
print(f"Mutual Consistency: {result.mutual_acc:.3f}")
```

```python
# Cell 5: Run E2 - Homogeneous Mutual Consistency
from mc_irf.experiments.runner import run_experiment

print("Running E2: Homogeneous Mutual Consistency (local)")
print("Generator: phi-3.5-mini-gguf, Discriminator: phi-3.5-mini-gguf")

result = run_experiment("E2", environment="local")

print(f"\n=== E2 Complete ===")
print(f"Generator Accuracy: {result.gen_acc:.3f}")
print(f"Discriminator Consistency: {result.disc_acc:.3f}")
print(f"Mutual Consistency: {result.mutual_acc:.3f}")
```

```python
# Cell 6: Run E0_local - Law Calibration (2 pairs)
from mc_irf.experiments.runner import run_experiment

print("Running E0_local: MCRF Law Calibration (local)")
print("Pairs: Φ→Φ, Φ→Qwen7B")

result = run_experiment("E0_local", environment="local")

print(f"\n=== E0_local Complete ===")
print(f"Generator Accuracy: {result.gen_acc:.3f}")
print(f"Discriminator Consistency: {result.disc_acc:.3f}")
print(f"Mutual Consistency: {result.mutual_acc:.3f}")
if result.rho:
    print(f"Error Correlation (ρ): {result.rho:.3f}")
if result.beta:
    print(f"Filtering Efficiency (β): {result.beta:.3f}")
```

```python
# Cell 7: Run E5_local - Cross-Architecture Validation
from mc_irf.experiments.runner import run_experiment

print("Running E5_local: Cross-Architecture (local)")
print("Generator: phi-3.5-mini-gguf, Discriminator: qwen2.5-7b-gguf")

result = run_experiment("E5_local", environment="local")

print(f"\n=== E5_local Complete ===")
print(f"Generator Accuracy: {result.gen_acc:.3f}")
print(f"Discriminator Consistency: {result.disc_acc:.3f}")
print(f"Mutual Consistency: {result.mutual_acc:.3f}")
if result.rho:
    print(f"Error Correlation (ρ): {result.rho:.3f}")
if result.beta:
    print(f"Filtering Efficiency (β): {result.beta:.3f}")
```

```python
# Cell 8: Aggregate Local Results & Fit MCRF Law
from mc_irf.theory.mcrf_law import MCRFLaw
from mc_irf.experiments.metrics import MCRFMetrics
import yaml
from pathlib import Path

print("Loading local experiment metrics...")

exp_ids = ["E1", "E2", "E0_local", "E5_local"]
all_metrics = []

for exp_id in exp_ids:
    metrics_path = f"experiments/results_local/{exp_id}/mcrf_metrics.yaml"
    if Path(metrics_path).exists():
        with open(metrics_path) as f:
            data = yaml.safe_load(f)
        # Convert to ExperimentMetrics
        from mc_irf.theory.mcrf_law import ExperimentMetrics
        if data.get("disc_acc"):
            all_metrics.append(ExperimentMetrics(
                exp_id=data["exp_id"],
                generator=data["generator"],
                discriminator=data["discriminator"] or "",
                nop_type=data["nop_type"],
                gen_acc=data["gen_acc"],
                disc_acc=data["disc_acc"],
                mutual_acc=data["mutual_acc"],
                sc_acc=data.get("sc_acc"),
                rho=data.get("rho"),
            ))

print(f"Loaded {len(all_metrics)} experiment metrics with discriminators")

# Fit MCRF law
law = MCRFLaw()
params = law.fit_law(all_metrics)

print(f"\n=== MCRF Law Fit (Local) ===")
print(f"β per NoOp type: {params.beta}")
print(f"β 95% CI: {params.beta_ci}")
print(f"ρ per NoOp type: {params.rho}")
print(f"ρ 95% CI: {params.rho_ci}")
print(f"R²: {params.r2:.4f}")
print(f"MSE: {params.mse:.4f}")
```
