```python
# Cell 2: Download Models (Run Once)
import subprocess
import yaml

with open('configs/models_h200.yaml') as f:
    config = yaml.safe_load(f)

models = config.get('models', {})
download_priority = config.get('download_priority', {})

print("Phase 1: Essential Models")
for key in download_priority.get('phase1_essential', []):
    if key in models:
        spec = models[key]
        result = subprocess.run(["huggingface-cli", "download", spec["hf_id"]], capture_output=True, text=True)
        if result.returncode == 0:
            print(f"  ✅ {spec['hf_id']}")
        else:
            print(f"  ❌ {spec['hf_id']}: {result.stderr[:100]}")

print("\nPhase 2: Scale Models (AWQ)")
for key in download_priority.get('phase2_scale', []):
    if key in models:
        spec = models[key]
        result = subprocess.run(["huggingface-cli", "download", spec["hf_id"]], capture_output=True, text=True)
        if result.returncode == 0:
            print(f"  ✅ {spec['hf_id']}")
        else:
            print(f"  ❌ {spec['hf_id']}: {result.stderr[:100]}")

print("\nPhase 3: Analysis Models (HF 8-bit)")
for key in download_priority.get('phase3_analysis', []):
    if key in models:
        spec = models[key]
        result = subprocess.run(["huggingface-cli", "download", spec["hf_id"]], capture_output=True, text=True)
        if result.returncode == 0:
            print(f"  ✅ {spec['hf_id']}")
        else:
            print(f"  ❌ {spec['hf_id']}: {result.stderr[:100]}")
```

```python
# Cell 3: Validate Model Registry
from mc_irf.models.registry import get_registry

reg = get_registry('configs/models_h200.yaml')
print("Available models:", reg.list_models())
print("\nFP16 models:", reg.list_models(backend='vllm', max_size_gb=17))
print("AWQ models:", [m for m in reg.list_models(backend='vllm') if 'awq' in m])
print("HF models:", reg.list_models(backend='hf'))

# Check feasible pairs
pairs = reg.get_feasible_pairs(max_size_gb=16.0)
print(f"\nFeasible pairs (16GB limit): {len(pairs)}")
for gen, disc in pairs[:10]:
    print(f"  {gen} -> {disc}")
if len(pairs) > 10:
    print(f"  ... and {len(pairs) - 10} more")
```

```python
# Cell 4: Test vLLM Load (Phi-3.5-mini)
from mc_irf.models.registry import get_registry

reg = get_registry('configs/models_h200.yaml')

print("Testing vLLM load: phi-3.5-mini")
model = reg.load('phi-3.5-mini')
print(f"Loaded: {model.get_model_info()}")

# Quick generation test
outputs = model.generate_single("2+2=", {"temperature": 0.0, "max_tokens": 10, "n": 1})
print(f"Generation: {outputs[0].text.strip()}")

reg.unload('phi-3.5-mini')
print("✅ Unloaded successfully")
```

```python
# Cell 5: Test Sequential Loading (Generator -> Discriminator)
from mc_irf.models.registry import get_registry, SequentialLoader

reg = get_registry('configs/models_h200.yaml')

print("Testing SequentialLoader: phi-3.5-mini -> qwen2.5-7b-awq")

with SequentialLoader(reg, 'phi-3.5-mini', 'qwen2.5-7b-awq') as loader:
    gen = loader.load_generator()
    print(f"Generator loaded: {gen.get_model_info()}")
    
    disc = loader.load_discriminator()
    print(f"Discriminator loaded: {disc.get_model_info()}")
    
    # Quick test
    gen_out = gen.generate_single("2+2=", {"temperature": 0.0, "max_tokens": 10, "n": 1})
    disc_out = disc.generate_single("Problem: 2+2=4\nSolution: 4", {"temperature": 0.0, "max_tokens": 10, "n": 1})
    print(f"Gen: {gen_out[0].text.strip()}")
    print(f"Disc: {disc_out[0].text.strip()}")

print("✅ Sequential loading works!")
```

```python
# Cell 6: Validate Data Generation
from mc_irf.data.gsm_symbolic import generate_problem_set, CORE_TEMPLATES, save_problems

print("Generating test problems...")
problems = generate_problem_set(
    templates=CORE_TEMPLATES,
    num_problems=5,
    seeds=[42, 123],
    nop_types=["quantity", "comparison", "filter", "discount"],
)

print(f"Generated {len(problems)} problems")
clean = [p for p in problems if p.nop_type is None]
print(f"  Clean: {len(clean)}")
for nt in ["quantity", "comparison", "filter", "discount"]:
    print(f"  {nt}: {len([p for p in problems if p.nop_type == nt])}")

# Save for experiments
save_problems(problems, "data/processed/h200_test_problems.jsonl")
print("Saved to data/processed/h200_test_problems.jsonl")
```

```python
# Cell 7: Run Pipeline Validation
from mc_irf.experiments.runner import run_experiment

# Quick test with minimal config - E1 with 1 problem, 1 seed, 4 rollouts
# This validates the full pipeline end-to-end

print("Running minimal E1 test (1 problem, 1 seed, 4 rollouts)...")
result = run_experiment("E1", environment="h200")

print(f"\n=== E1 Test Complete ===")
print(f"Generator Accuracy: {result.gen_acc:.3f}")
print(f"Mutual Consistency: {result.mutual_acc:.3f}")
```
